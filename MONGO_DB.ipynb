```python
from pymongo import MongoClient
client = MongoClient('mongodb://localhost:27017/')
print("Connected to MongoDB successfully!")
```

```python
mydb = client['mydatabase']
mycollection = mydb['myTable']

record = {
    "title": "MongoDB and Python",
    "description": "Mongo is noSQL database",
    "tags": ["mongodb","database","NoSQL"],
    "viewers": 104
}

result = mycollection.insert_one(record)
print("Inserted ID:", result.inserted_id)
```

```python
for i in mycollection.find({"title": "MongoDB and Python"}):
    print(i)
```

```python
count = mycollection.count_documents({"title": "MongoDB and Python"})
print("Number of documents with title 'MongoDB and Python':", count)
```

```python
from pymongo import MongoClient

try:
    conn = MongoClient('mongodb://localhost:27017/')
    print("Connected to MongoDB successfully!")
except Exception as e:
    print("Error connecting to MongoDB:", e)

db = conn['mydatabase']
collection = db['myTable']
for record in collection.find({"title": "MongoDB and Python"}):
    print(record)
```
