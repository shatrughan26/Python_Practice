```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
def f(x):
    return x**2 + 4*x + 4
def df(x):
    return 2*x + 4
def gradient_descent(starting_point, learning_rate, iterations):
    x = starting_point
    for i in range(iterations):
        x = x - learning_rate * df(x)
        print(f"Iteration {i + 1}: x = { x :.4f}, f(x) = {f(x):.4f}")
    return x

starting_point = 0
learning_rate = 0.1
iterations = 10

minimum = gradient_descent(starting_point, learning_rate,iterations)
print(f"\nLocal minimum occurs at x = {minimum:.4f},f(x) = {f(minimum):.4f}")
```

```python
x_vals = np.linspace(-10, 2 ,100)
y_vals = f(x_vals)
plt.plot(x_vals,y_vals,label= "f(x) = x^2 + 4x + 4")
plt.scatter(minimum, f(minimum),color = 'red', label = "Local minimum")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Gradient Descent Visualization")
plt.legend()
plt.show()
```

```python
df = pd.read_csv("Studnent_marks.csv")
df
```

```python
z = df['math'].to_numpy()
y = df['cs'].to_numpy()
```

```python
print(type(z))
print(z)
```

```python
m = 0 #slop
b = 0 #intercept
learning_rate = 0.0001
iterations = 1000
a = len(z)

for i in range(iterations):
    y_pred = m*z + b

    dm = (-2/a) * np.sum(z * (y - y_pred))
    db = (-2/a) * np.sum(y - y_pred)

    m = m - learning_rate * dm
    b = b - learning_rate * db

print("m:", m)
print("b:", b)
```

```python
import matplotlib.pyplot as plt

plt.scatter(z,y)

y_line = m * z + b
plt.plot(z, y_line)

plt.xlabel("Maths Marks")
plt.ylabel("CS Marks")
plt.title("linear Regression using Gradient Descent")

plt.show()
```
