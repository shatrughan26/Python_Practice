```python
f = open("geek.txt", "r")
print("Filename: ", f.name)
print("Mode: ", f.mode)
print("Is Closed: ", f.closed)
f.close()

print("Is Closed: ", f.closed)
```

```python
with open("geek.txt", "w") as file:
    file.write("Hello, Python!\n")
    file.write("File handling in Python is easy.\n")

print("File written successfully.")
```

```python
f= open("geek.txt", "r")
content = f.read()
print(content)
f.close()
```

```python
with open("geek.txt", "r") as file:
    content = file.read()
    print(content)
```
