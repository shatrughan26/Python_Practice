```python
import pandas as pd
df = pd.read_csv("car_price_data.csv")
df.head()
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
plt.scatter(df['Mileage'],df['Sell Price($)'])
```

```python
plt.scatter(df['Age(yrs)'],df['Sell Price($)'])
```

```python
x = df[['Age(yrs)','Mileage']]
y = df['Sell Price($)']
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x,y, test_size = 0.3)
```

```python
len(x)
```

```python
from sklearn.linear_model import LinearRegression
clf = LinearRegression()
clf.fit(X_train, y_train) 
```

```python
X_test
```

```python
predicted_price = clf.predict(X_test)
predicted_price
```

```python
y_test
```

```python
import matplotlib.pyplot as plt

# Make sure data is 1D
y1 = predicted_price.flatten()
y2 = y_test.values.flatten()

plt.hist(y1, bins=10, alpha=0.5, label='Predicted Price')
plt.hist(y2, bins=10, alpha=0.5, label='Actual Price')

plt.xlabel("Price")
plt.ylabel("Frequency")
plt.title("Histogram of Predicted vs Actual Prices")

plt.legend()
plt.show()
```

```python
clf.score(X_test,y_test)
```
