```python
im = np.eye(3)
da = np.diag([1, 2, 3])
a0 = np.zeros_like(da)
a1 = np.ones_like(da)

print("Identity matrix:\n", im)
print("diagonal matrix:\n", da)
print("zeros matrix:\n", a0)
print("ones matrix:\n", a1)
```
